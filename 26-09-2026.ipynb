### Dict

```python
students = { 
                101 : ["Monal", "abc@gmail.com", 80],
                102 : ["Richard", "abc@gmail.com", 80],
                '103' : ["Rishabh", "abc@gmail.com", 80],
                101 : ["Mahesh", "abc@gmail.com", 80],
        }
```

```python
# how to get the value from dict

# Sol.1 -> list kind syntax
students[102]
```

```python
# Sol.1 -> list kind syntax
students[101]
```

```python
# Sol.1 -> list kind syntax
students[105]
```

```python
# Sol.2 using .get
students.get(102, "Value not present")
```

```python
# Sol.2 using .get
students.get(105, "Value not present")
```

```python
# getting all keys and their values

students.keys()
```

```python
students.values()
```

```python
students.items()
```

```python
# adding values

# students[key] = value

students[104] = ['shyamu', 'abc@gmail.com', 80]
```

```python
students
```

```python
students[104] = []
```

```python
students
```

```python
# students.pop(key)
students.pop(104)
```

```python
students
```

```python
# list syntax style key : value removal

del students['103']
```

```python
students
```

```python
# adding values - method-2 (.update())

students.update({
                    103 : ['Rishabh', 'abc@gmail.com', 80], 
                    104: ['shyamu', 'abc@gmail.com', 80]
                })
```
