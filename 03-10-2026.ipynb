```python
def cube_and_add(a,b):
    # print(a**3 + b**3)
    return a**3 + b**3

def inverse(c):
    # print(c**-1)
    return c**-1

def formula(a,b):
    print("calculation started")
    value_1 = cube_and_add(a,b)
    value_2 = inverse(value_1)
    
    return value_2

def formula_v2(a,b):
    print("calculation started")

    value_1 = inverse(cube_and_add(a,b))
    
    return value_1

def formula_v3(a,b):
    print("calculation started")
    
    return inverse(cube_and_add(a,b))

```

after positional arguments (*args), python will look for keyword arguments

```python
def add_v5(a, b, *values, **new_value):
    # *values --> catch values as elements of tuple
    # **new_value -> catch values as elements of dict
    print(a)
    print(b)
    print(values)
    print(new_value)
```

```python
add_v5(1,2,3,4,5,6,7,8, c=2, batch="ultimate data science 3.0")
```

```python
add_v5(1,2,3,4,5,6,7,8, c=2, batch="ultimate data science 3.0",9)
```

```python
def calculation(a,b,**action_todo):
    answer = {}
    check_action = list(action_todo.keys())
    # print(check_action)
    if 'add' in check_action:
        answer['add'] = a+b
    if 'subtract' in check_action:
        answer['subtract'] = a-b
    if 'multiply' in check_action:
        answer['multiply'] = a*b
    if 'divide' in check_action:
        answer['divide'] = a/b
    return answer
```

```python
calculation(10,20, add=True, subtract=True)
```

```python
calculation(10,20, divide=True, subtract=True)
```

```python
calculation(10,20, multiply=True, subtract=True, c=3)
```
