# Python Operators

### Arithmetic

```python
val_1 = 100
val_2 = 3

print("Addition:", val_1 + val_2)
print("Subtraction:", val_1 - val_2)
print("Multiplication:", val_1 * val_2)
print("Division:", val_1 / val_2) # returns float
print("Floor-division:", val_1 // val_2) # return integer
print("Modulus:", val_1 % val_2)
print("Exponential:", val_1 ** val_2)
```

```python
print(val_1, val_2, 3, "hello")
```

### Relational/Comparison Operator

Operators or return true or false

```python
val_1 = 100
val_2 = 3

print(val_1 > val_2)
print(val_1 < val_2)
print(val_1 >= val_2) # equal or greater
print(val_1 <= val_2) # equal or smaller
print(val_1 != val_2)
print(val_1 == val_2)
```
