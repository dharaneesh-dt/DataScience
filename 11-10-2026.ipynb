```python
def try_except_func_1(divide_by):
    try : 
        return 10 / divide_by
    except ZeroDivisionError:
        print("Inside ZeroDivisionError except block")
        return 0
    except FileNotFoundError:
        # logic to handle error of file not found
        print("Inside FileNotFoundError except block")
        return 0
    except Exception as e:
        print("Inside Base Exception except block")
        return e
```

# CustomException

```python
# definiting your custom exception

class AgeValueShouldBeNumber(Exception):
    pass
```

```python
def go_to_club(age, has_id=False):

    if type(age) == int:
        if age >= 18:
            if has_id:
                return True
            else :
                return False
        else :
            return False
    else :
        raise AgeValueShouldBeNumber("Age should be a number, String or float or any other data type is not allowed.")
```

```python
age = float(input("Enter your age : "))
try : 
    value = go_to_club(age, True)
except AgeValueShouldBeNumber as e:
    print(e)
    age = int(age)
    value = go_to_club(age, True)
except Exception as e:
    value = 0
    print("Default Exception")
print(value)
```
