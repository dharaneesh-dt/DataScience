```python
# string formatting
table_of = 2
multiplied_by = 424672472
answer = table_of*multiplied_by
# 2 x 3 = 6
string_representation = str(table_of) + " X " + str(multiplied_by) + " = " + str(answer)
print(string_representation) #1
print(table_of, "X", multiplied_by, "=", answer)#2
```

```python
# using f-strings (only available from python 3.6+)
string_representation = f"{table_of} X {multiplied_by} = {answer}"
print(string_representation)
```

```python
# using .format (only available from python 3.6+)
string_representation_4 = "{} X {} = {}".format(table_of,multiplied_by,answer)
print(string_representation_4)
```
