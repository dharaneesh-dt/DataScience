# 4 pillar of OOP

```python
class Animal:

    def __init__(self, legs):
        self.leg_count = legs

    def how_many_legs(self):
        print(f"Has {self.leg_count} legs.")

    def eat(self):
        print("I can Eat!")
```

```python
class Dog(Animal): # Parent --> Animal, Dog is child class inheriting traits of Animal

    def __init__(self, name, legs):
        self.name = name
        super().__init__(legs)

    def bark(self):
        print(f"{self.name} says woof!")

    def parant_var(self):
        print(self.leg_count)

class Cat(Animal): # Parent --> Animal, Cat is child class inheriting traits of Animal

    def __init__(self, name):
        self.name = name

    def meow(self):
        print(f"{self.name} says meow!")
```

```python
dog_1 = Dog("ellie", 4)
dog_1.bark()
```

```python
dog_1.how_many_legs()
```

```python
dog_1.parant_var()
```

```python
dog_1.eat()
```

```python
cat_1 = Cat("Garfield")
cat_1.meow()
```

```python
cat_1.eat()
```

# Composition

```python
class Bird:

    def fly(self):
        print("Can fly!")
```

```python
class Eat:
    def eat(self):
        print("Can eat!")
```

```python
class Swim:
    def swim(self):
        print("Can swim!")
```

```python
class Penguin(Eat):
    def __init__(self, swim_obj):
        self.swim_obj = swim_obj

    def swim(self):
        self.swim_obj.swim()
```

```python
swim_obj = Swim()
```

```python
swim_obj.swim()
```

```python
penguin_obj = Penguin(swim_obj)
```

```python
penguin_obj.eat()
```

```python
penguin_obj.swim()
```

# Example 2 -> Composition

```python
# component 1 : Engine
class Engine:

    def __init__(self, housepower):
        self.housepower = housepower

    def start(self):
        print(f"Engine with {self.housepower} is started.")

# component 2 : Sound System
class SoundSystem:

    def __init__(self, brand):
        self.brand = brand

    def play_music(self):
        print(f"Playing music through {self.brand} speakers.")
```

```python
class Car:

    def __init__(self, company_name, model_name, engine_obj, sound_obj):
        self.company_name = company_name
        self.model_name = model_name
        # Composition > injecting independent objects into the car
        self.engine = engine_obj
        self.sound_system = sound_obj

    def drive(self):
        print(f"Starting your {self.company_name} - {self.model_name}")
        self.engine.start()
        self.sound_system.play_music()
```

```python
v8_engine = Engine(1000)
bose_sound_system = SoundSystem("Bose")
```

```python
car_1 = Car("Ford", "Mustang GT", v8_engine, bose_sound_system)
```

```python
car_1.drive()
```

# Encapsulation

```python
class Car:

    def __init__(self, brand):
        self.brand = brand # public
        self._ac_voltage = 30 # protected
        self.__current_gps_coordinate = "LA 1.23 LO 4.56" # private

    def tell_brand(self):
        print(self.brand)

    def run_ac(self):
        if self.check_ac():
            print("Starting AC")

    def check_ac(self):
        if self._ac_voltage > 30:
            return False
        else :
            return True

    def accessing_gps_coordinate(self, password):
        if password == 'abc123':
            print(self.__current_gps_coordinate)

    def update_coordinate(self):
        # logic only
        self.__current_gps_coordinate = "LOGIC TO UPDATE COORDINATE"
```

```python
car_obj_1 = Car("Toyota")
```

```python
car_obj_1.brand
```

```python
car_obj_1.tell_brand()
```

```python
car_obj_1._ac_voltage = 100 # protected
```

```python
car_obj_1._ac_voltage
```

```python
car_obj_1.check_ac()
```

```python
car_obj_1.__current_gps_coordinate
```

```python
car_obj_1._Car__current_gps_coordinate
```

```python
car_obj_1.accessing_gps_coordinate("abc123")
```

```python
# name mangling example

class Machine:
    def __init__(self):
        self.__status = "Machine ready"

    def get_machine_status(self):
            return self.__status

class Printer(Machine):
    def __init__(self):
        super().__init__()

        self.__status = "Printer ready"

    def get_printer_status(self):
            return self.__status
```

```python
my_printer_obj = Printer()
```

```python
my_printer_obj.get_machine_status()
```

```python
my_printer_obj.get_printer_status()
```

```python
my_printer_obj._Machine__status
```
